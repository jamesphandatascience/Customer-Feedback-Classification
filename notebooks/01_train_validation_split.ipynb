```python
import pandas as pd
import sqlite3
from sklearn.model_selection import train_test_split
```

```python
X = pd.read_csv("../data/X_train.csv")
y = pd.read_csv("../data/y_train.csv")

conn = sqlite3.connect(":memory:")

X.to_sql("X", conn, index=False, if_exists="replace")
y.to_sql("y", conn, index=False, if_exists="replace")

X_train, X_val, y_train, y_val = train_test_split(X, 
                                                  y, 
                                                  test_size=0.2, 
                                                  stratify=y, 
                                                  random_state=42)

X_train.to_sql("X_train", conn, index=False, if_exists="replace")
y_train.to_sql("y_train", conn, index=False, if_exists="replace")
X_val.to_sql("X_val", conn, index=False, if_exists="replace")
y_val.to_sql("y_val", conn, index=False, if_exists="replace")
```

```python
# Examine label distribution in training data
query = '''
SELECT label, COUNT(*) AS count 
FROM y
GROUP BY label
ORDER BY count DESC'''

label_distribution = pd.read_sql(query, conn)
print(label_distribution.to_string(index=False))
```

```python
# Examine label distribution in validation data
query = '''
SELECT label, COUNT(*) AS count 
FROM y_val
GROUP BY label
ORDER BY count DESC'''

validation_label_distribution = pd.read_sql(query, conn)
print(validation_label_distribution.to_string(index=False))
```

Note that the validation set has counts for all labels which is what we want.

```python
# Export features
X_train.to_csv("../data/X_train_split.csv", index=False)
X_val.to_csv("../data/X_validation.csv", index=False)

# Export labels
y_train.to_csv("../data/y_train_split.csv", index=False)
y_val.to_csv("../data/y_validation.csv", index=False)
```
